```python
from getpass import getpass

from qiskit_ibm_runtime import QiskitRuntimeService

token = getpass("IBM Quantum API key: ")
instance = input("Instance CRN or name (press Enter to skip): ").strip() or None

QiskitRuntimeService.save_account(token=token, instance=instance, channel="ibm_quantum_platform",
                                  name="qsegment", overwrite=True)
del token

service = QiskitRuntimeService(name="qsegment")
print("saved account name: qsegment")
print("backends you can use:", [b.name for b in service.backends()])
```
